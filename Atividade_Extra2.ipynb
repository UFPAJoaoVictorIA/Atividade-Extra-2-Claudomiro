```markdown
# Análise de Capacidade de Memória DRAM

Este notebook apresenta o cálculo detalhado da capacidade de armazenamento de um chip de memória dinâmica (DRAM), considerando a multiplexação de endereços (Linha e Coluna) e a estrutura física dos pinos.

### Parâmetros Analisados:
* **Pinos de Endereço (A0-A10):** 11 pinos físicos.
* **Arquitetura:** Multiplexada (22 bits de endereço total).
* **Palavra de Dados:** 1 bit (D).
```

````python
# 1. Parâmetros físicos do diagrama
pinos_endereco_fisico = 11  # Pinos A0 até A10
tamanho_palavra_bits = 1    # Pino D (1 bit por endereço)

# 2. Resolução da Multiplexação (DRAM)
bits_endereco_total = pinos_endereco_fisico * 2

# 3. Cálculo do total de endereços endereçáveis (2^n)
total_enderecos = 2 ** bits_endereco_total

# 4. Cálculo da capacidade de armazenamento
capacidade_total_bits = total_enderecos * tamanho_palavra_bits

# 5. Conversões de unidades (dividindo por 1024 sucessivamente)
capacidade_kbits = capacidade_total_bits / 1024
capacidade_mbits = capacidade_kbits / 1024

# 6. Exibindo os Resultados
print("--- RESULTADOS DA ANÁLISE DO CHIP ---")
print(f"Pinos físicos de endereço: {pinos_endereco_fisico} (A0-A10)")
print(f"Bits totais de endereço: {bits_endereco_total} bits (Linha + Coluna)")
print(f"Total de posições de memória: {total_enderecos}")
print(f"Tamanho da palavra de dados: {tamanho_palavra_bits} bit\n")

print("--- CAPACIDADE ---")
print(f"Em Bits: {capacidade_total_bits}")
print(f"Em Kbits: {capacidade_kbits} Kbits")
print(f"Em Mbits: {capacidade_mbits} Mbits\n")

print(f"Conclusão: O cálculo bate perfeitamente com a especificação")
print(f"do diagrama de {capacidade_kbits} K x {tamanho_palavra_bits} ({capacidade_mbits} Mbits)!")
````

````python
# ==========================================
# CÁLCULO DE CAPACIDADE DO CHIP DE MEMÓRIA
# ==========================================

# 1. Definição dos parâmetros físicos do diagrama
pinos_endereco_fisico = 11  # Pinos A0 até A10
tamanho_palavra_bits = 1    # Pino D (1 bit por endereço)

# 2. Resolução da Multiplexação (DRAM)
# O endereço real é a soma dos bits de linha (RAS) e coluna (CAS)
bits_endereco_total = pinos_endereco_fisico * 2

# 3. Cálculo do total de endereços endereçáveis (2^n)
total_enderecos = 2 ** bits_endereco_total

# 4. Cálculo da capacidade de armazenamento
capacidade_total_bits = total_enderecos * tamanho_palavra_bits

# 5. Conversões de unidades (dividindo por 1024 sucessivamente)
capacidade_kbits = capacidade_total_bits / 1024
capacidade_mbits = capacidade_kbits / 1024

# ==========================================
# SAÍDA DOS RESULTADOS FORMATADOS
# ==========================================
print("-" * 50)
print(" RESULTADOS DA ANÁLISE DO CHIP".center(50))
print("-" * 50)
print(f"Pinos físicos de endereço:   {pinos_endereco_fisico} (A0-A10)")
print(f"Bits totais de endereço:     {bits_endereco_total} bits (Linha + Coluna)")
print(f"Total de posições de mem.:   {total_enderecos:,.0f}".replace(',', '.'))
print(f"Tamanho da palavra de dados: {tamanho_palavra_bits} bit\n")

print("--- Capacidade ---")
print(f"Em Bits:   {capacidade_total_bits:,.0f}".replace(',', '.'))
print(f"Em Kbits:  {capacidade_kbits:,.0f} Kbits".replace(',', '.'))
print(f"Em Mbits:  {capacidade_mbits:.0f} Mbits\n")

# Validação final em relação ao texto do diagrama
print(f"Conclusão: O cálculo bate perfeitamente com a especificação")
print(f"do diagrama de {capacidade_kbits:,.0f} K x {tamanho_palavra_bits} ({capacidade_mbits:.0f} Mbits)!".replace(',', '.'))
print("-" * 50)
````
